## Búsqueda en strings

```python
texto = "el clima de talca\nes agradable en primavera"
print("talca" in texto)
print("santiago" in texto)
print(texto)
```

```python
print(texto.find("talca"))
print(texto.find("santiago"))
```

```python
# Lo MINIMO (funciona, falla en estetica, lectura y resultado)
try:
  print(texto.index("talca"))
except ValueError:
  print("La palabra no se encuentra en el texto")
```

## Busqueda elegante

```python
# Texto separado vs string unido
print(texto)
print(repr(texto))
```

```python
filas = texto.split("\n")
print(filas)
```

```python
# FILAS
for fila in filas:
  print(fila)
```

```python
# ¿Que numero de fila?
for numero_fila, fila in enumerate(filas):
  print(f"Esta es la fila {numero_fila + 1}: contenido de la fila {fila}")
```

```python
palabra = "primavera"
if palabra in texto:
  print(f"La palabra {palabra} se encuentra en el texto")
else:
  print(f"La palabra {palabra} NO se encuentra en el texto")
```

```python
# Busqueda completa
for fila in filas:
  if palabra in fila:
    print(f"La palabra {palabra} se encuentra en la fila {filas.index(fila) + 1} en la columna {fila.find(palabra)}")
```

## Repaso general

```python
# ¿Si no existe se deja en blanco?
try:
  with open("texto.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
except FileNotFoundError:
  print("El archivo no existe.")
  contenido = ""

# Separar palabras
palabras = contenido.split()

# Contar de manera eficiente (no obligatoria)
vocales = "aeiou"
conteo_vocales = {v: 0 for v in vocales}
for letra in contenido.lower():
  if letra in vocales:
    conteo_vocales[letra] += 1

print("Conteo de vocales:")
for v, c in conteo_vocales.items():
  print(f"  {v}: {c}")

if palabras:
  mas_repetida = max(set(palabras), key=palabras.count)
  print(f"Palabra más repetida: {mas_repetida} ({palabras.count(mas_repetida)} veces)")
```
