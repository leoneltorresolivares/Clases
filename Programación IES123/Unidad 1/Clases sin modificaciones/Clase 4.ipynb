## Robustez básica

```python
# Revisar SIEMPRE
numero = input("Ingrese un número: ")
numero = numero.replace(",",".")
numero = float(numero)
print(numero)
```

```python
numero = input("Ingrese un número: ")
numero = numero.replace(",", ".")

try:
  numero = float(numero)
  print(f"El {numero} pudo transformarse a número")
except ValueError:
  print(f"El {numero} no pudo transformarse a número")
```

```python
def validar_numero():
  while True:
    dato = input("Ingrese su edad: ")
    dato = dato.replace(",", ".")

    try:
      return float(dato)

    except ValueError:
      print("Debe ingresar un número valido.")
```

```python
edad = validar_numero()
print(f"Su edad es {edad} años.")
```

### ¿Y si el archivo no abre?

```python
# Usar try con el archivo .txt
```

```python
va = input("Ingrese el nombre del archivo: ")
va = va + ".txt"
```

```python
while True:
  try:
    with open(va, "r", encoding="utf-8") as archivo:
        texto = archivo.read()
        break

  except FileNotFoundError:
    print("El archivo no existe. Carguelo manual")
```

```python
# robustez
print(texto)
```
