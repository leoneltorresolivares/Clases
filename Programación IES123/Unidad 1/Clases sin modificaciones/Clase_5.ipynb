## Contar vocales

```python
vocales = "aeiou"
# diccionario
conteo = {v: 0 for v in vocales}
print(conteo)
```

```python
print(texto)
```

```python
# Mostrar el diccionario
print(conteo)
```

```python
texto = "estadio"
```

```python
print(vocales)
```

```python
cont = 0
for letra in texto:
  if letra in vocales:
    conteo[letra] += 1
    cont = cont + 1

for v, c in conteo.items():
  print(f"La vocal '{v}' aparece {c} veces")

print(f"Hay {cont} vocales")
```

```python
consonantes = len(texto) - cont
print(f"Hay {consonantes} consonantes")
```

## Comparación de strings

```python
nombre1 = "Ana"
nombre2 = "Beatriz"
print(nombre1 == nombre2)
print(nombre1 < nombre2)
print(nombre1 > nombre2)
```

```python
# Orden lexico
nombres = ["Beatriz", "ana", "Carlos", "andrea"]
print(sorted(nombres))
```

```python
# Comparación insensible a mayúsculas
print(sorted(nombres, key=str.lower))
```

## Conversión de caracteres y números

```python
print(ord("a"))
print(ord("A"))
print(chr(97))
print(chr(65))
```

```python
# Mini tabla ASCII de las primeras letras
for letra in "ABCDE":
  print(f"{letra} --- {ord(letra)}")
```

```python
dato = "123"
print(dato.isdigit())

dato2 = "12a"
print(dato2.isdigit())
```

```python
# Contar dígitos en un input, caracter por caracter
dato = input("Ingrese un texto: ")
digitos = 0
for c in dato:
  if c.isdigit():
    digitos += 1
print(f"El texto ingresado tiene {digitos} dígitos")
```

```python
# Validar antes de convertir, aplicado al sistema de notas
def validar_nota(mensaje):
  while True:
    dato = input(mensaje)
    dato = dato.replace(",", ".")
    if dato.replace(".", "", 1).isdigit():
      return float(dato)
    else:
      print("Debe ingresar un número válido.")
```

```python
nota = validar_nota("Ingrese su nota: ")
print(f"Nota ingresada: {nota}")
```

## Robustez carga de archivos

```python
while True:
  va = input("Ingrese el nombre del archivo sin la extensión del mismo: ")
  va = va + ".txt"
  try:
    with open(va, "r", encoding="utf-8") as archivo:
        texto = archivo.read()
        print("Archivo cargado correctamente!")
        break

  except FileNotFoundError:
    print("El archivo no existe.\nVerifique el nombre ingresado")
```
