```python
# Recorrido
nota = float(input("Ingrese su nota: "))
if nota < 1.0 or nota > 7.0:
  print("Nota invalida")
elif nota >= 4.0:
  print("Aprobado")
elif nota >= 3.0 and nota <= 4.0:
  print("Morado")
else:
  print("Reprobado")
```

```python
notas = []
while True:
  nota = input("Ingrese una nota (k para terminar): ")
  if nota == "k":
    break
  else:
    try:
      nota = float(nota)
      if nota < 1.0 or nota > 7.0:
        print(f"{nota} : Nota invalida")
      else:
        notas.append(nota)
    except ValueError:
      print(f"El {nota} no es un número")
print(notas)
```

# Block de notas

```python
with open("Bloc_notas_test.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()
print(lineas)
```

```python
while True:
  print("1: Visualizar información del archivo")
  print("2: Agregar información al archivo")
  print("3: Salir del programa")
  opcion = input("Ingrese su opción: ")

  if opcion == "1":
    print(lineas)
    input("Ingrese ENTER para volver al menu.")

  elif opcion == "2":
    info_new = input("Ingrese la información que desea agregar: ")
    lineas.append(info_new + "\n\n")
    print(lineas)
    input("Ingrese ENTER para volver al menu.")

  elif opcion == "3":
    break
  else:
    print("Opción inválida")
    input("Ingrese ENTER para volver al menu.")


with open("Bloc_notas_test_v2.txt", "w", encoding="utf-8") as archivo:
    archivo.writelines(lineas)
```
