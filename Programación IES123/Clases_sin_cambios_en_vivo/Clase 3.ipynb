## Uso de funciones

```python
# funcion
def clasificar(nota):
  if nota >= 4.0:
    return "Aprobado"
  elif nota >= 3.0:
    return "Morado"
  else:
    return "Reprobado"
```

```python
print(clasificar(7.0))
print(clasificar(3))
```

```python
notas = [5.5, 2.0, 3.6, 4.0, 7.0]
# clasificar(notas)
```

```python
for nota in notas:
  print(f"{nota} ---", clasificar(nota))
```

```python
# Más detalle (parametros)
def clasificar(nota, aprobado, PAF):
  if nota >= aprobado:
    return "Aprobado"
  elif nota >= PAF:
    return "Examen"
  else:
    return "Reprobado"
```

```python
for nota in notas:
  print(f"{nota} ---", clasificar(nota, 4, 3.5))
```

```python
def leer_notas():
  notas = [] # O fuera? DEPENDE
  #nota = float(input("Ingresar su nota (k para break): "))
  while True:
    nota = input("Ingresar su nota (k para break): ")
    if nota != "k":
      try:
        nota = float(nota)
        notas.append(nota)
      except ValueError:
        print(f"La {nota} no pudo transformarse a número")
    else:
      break
  return notas

def porcentaje_aprobacion(notas):
  aprobados = sum(1 for n in notas if n >= 4.0)
  return round(100 * aprobados / len(notas), 5)
```

```python
notas = leer_notas()
for n in notas:
  print(n, "---", clasificar(n, 4.0, 3.5))
```

```python
print("Porcentaje de aprobación: ", porcentaje_aprobacion(notas), "%")
```
