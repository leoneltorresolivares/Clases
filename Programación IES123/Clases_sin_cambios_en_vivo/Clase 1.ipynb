```python
# lista
notas = [5.5, 6.0, 4.5, 7.0]
print(f"Las notas del estudiantes son: {notas}")
```

```python
# Suma y promedio
suma = sum(notas)
promedio = suma/len(notas)
print(f"La suma de las notas del estudiantes son {suma} y su promedio es {promedio}")
```

```python
# Interactivo
nombre = input("Ingrese su nombre: ")
print(f"Hola {nombre}")
notas = []
n = int(input("¿Cuantas notas vas a ingresar?: "))
for i in range(n):
  nota = float(input(f"Ingrese la nota {i+1}: "))
  notas.append(nota)
print(f"Las notas del estudiantes son: {notas}")
```

```python
min = min(notas)
max = max(notas)
print(f"La nota mínima del estudiante es {min} y la máxima es {max}")
```

```python
with open("Texto.txt", "r", encoding="utf-8") as archivo:
    palabras = archivo.read().split()

print(palabras)
len(palabras)
```

```python
print(len(palabras)) # Cantidad de palabras
print(palabras[0]) # Primera palabra
print(palabras[-1]) # Última palabra
print("python" in palabras) # ¿Existe la palabra?
print(palabras.count("Gears"))# Cantidad de x palabra
```
